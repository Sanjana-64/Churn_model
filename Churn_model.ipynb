```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler,LabelEncoder
import pickle
```

```python
df=pd.read_csv("/content/Churn_Modelling (1).csv")
```

```python
df.head()
```

```python
df=df.drop(['RowNumber','CustomerId','Surname'],axis=1)
```

```python
df.head()
```

```python
label_encoder_gender=LabelEncoder()
```

```python
df['Gender']=label_encoder_gender.fit_transform(df['Gender'])
```

```python
df.head()
```

```python
df['Geography'].value_counts()
```

```python
df['Geography']=label_encoder_gender.fit_transform(df['Geography'])
```

```python
df.head()
```

```python
x=df.drop('Exited',axis=1)
y=df['Exited']
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
scaler=StandardScaler()
```

```python
x_train=scaler.fit_transform(x_train)
```

```python
x_train
```

```python
x_test
```

```python
y_test
```

```python
x_test=scaler.fit_transform(x_test)
```

```python
x_test
```

```python
x_train.shape[1]
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
```

Model Training


```python
model=Sequential([
    Dense(64,activation='relu',input_shape=(x_train.shape[1],)), #input
    Dense(32,activation='relu'), #hidden layer
    Dense(1,activation='sigmoid') #output layer
])
```

```python
model.summary()
```

```python
import tensorflow
```

```python
opt=tensorflow.keras.optimizers.Adam(learning_rate=0.0)
loss=tensorflow.keras.losses.BinaryCrossentropy()
```

##compile the model


```python
model.compile(optimizer=opt, loss="binary_crossentropy", metrics=['accuracy'])
```

```python
model.fit(x_train,y_train,validation_data=(x_test,y_test), epochs=100)
```

```python
import pickle
```

```python
model.save('mymodel.keras')
```

Model Inference


```python
import numpy as np
from tensorflow.keras.models import load_model

# Load the saved model
loaded_model = load_model('mymodel.keras')

# Create some sample input data for prediction
# This should have the same 10 features as X_train
# Example values (replace with actual data you want to predict on)
sample_input = np.array([[10, 1, 2, 50, 30, 600, 2, 1, 1,100]]) # Example customer data

# Scale the sample input data using the same scaler used for training
scaled_sample_input = scaler.transform(sample_input)

# Make a prediction
prediction = loaded_model.predict(scaled_sample_input)

# The output is a probability, typically we convert it to a binary class (0 or 1)
binary_prediction = (prediction > 0.5).astype(int)

print(f"Prediction probability: {prediction[0][0]:.4f}")
print(f"Binary prediction (0=not exited, 1=exited): {binary_prediction[0][0]}")
```
